
# Data Quality and Validation in ETL - Assignment Solution



## SECTION A – THEORY

### Q1
Data quality refers to accuracy, completeness, consistency, and reliability of data in ETL. It is more than cleaning because it ensures data is usable for business decisions.

### Q2
Poor data quality leads to:
- Wrong KPIs
- Misleading dashboards
- Incorrect business decisions

### Q3
Duplicate data: Same record appearing multiple times  
Causes:
- Multiple data sources
- ETL pipeline errors
- System retries

### Q4
- Exact: identical records  
- Partial: some fields same  
- Fuzzy: similar but not identical

### Q5
Validation during transformation ensures errors are caught early before loading into final system.

### Q6
Business rules define expected values  
Example: Quantity > 0


## SECTION B – PRACTICAL

```python

import pandas as pd
import numpy as np

data = {
    "Txn_ID":[201,202,203,204,205,206,207,208],
    "Customer_ID":["C101","C102","C101","C103","C104","C105","C106","C101"],
    "Product_ID":["P11","P12","P11","P13","P14","P15","P16","P11"],
    "Quantity":[2,1,2,3,np.nan,1,1,2],
    "Txn_Amount":[4000,1500,4000,6000,2500,np.nan,1800,4000],
    "Txn_Date":["2025-12-01","2025-12-01","2025-12-01","2025-12-02","2025-12-02","2025-12-03",None,"2025-12-01"]
}

df = pd.DataFrame(data)
df
```

## Q7 Duplicate Keys

```python

duplicates = df.groupby(['Customer_ID','Product_ID','Txn_Date','Txn_Amount']).size().reset_index(name='Count')
duplicates[duplicates['Count']>1]
```

SQL Query:

```python

SELECT Customer_ID, Product_ID, Txn_Date, Txn_Amount, COUNT(*) as Count
FROM Sales_Transactions
GROUP BY Customer_ID, Product_ID, Txn_Date, Txn_Amount
HAVING COUNT(*) > 1;
```

## Q8 Referential Integrity

```python

customers = pd.DataFrame({
    "CustomerID":["C101","C102","C103","C104"]
})

invalid = df[~df['Customer_ID'].isin(customers['CustomerID'])]
invalid
```

SQL Query:

```python

SELECT st.Customer_ID
FROM Sales_Transactions st
LEFT JOIN Customers_Master cm
ON st.Customer_ID = cm.CustomerID
WHERE cm.CustomerID IS NULL;
```
